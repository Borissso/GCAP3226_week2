## 3. Green-light — load course data

The CSV is **in this repo**, next to the notebook. Survey wording is on the Week 2 slides.

```python
# Purpose of this cell:
# Load the course survey data (week2.csv) from THIS repo folder 
# "shape" = how many rows and columns the table has (like Excel: rows × columns).
# .head() shows the first few rows so you can peek at the data.

from pathlib import Path

DATA_FILE = "week2.csv"

def load_week2_data():
    candidates = [
        Path(DATA_FILE),
        Path.cwd() / DATA_FILE,
    ]
    for p in candidates:
        if p.is_file():
            print(f"Using data file: {p.resolve()}")
            return pd.read_csv(p)
    raise FileNotFoundError(
        f"Could not find {DATA_FILE}. Use a Codespace from your fork; "
        "do not use a path from your own computer."
    )

df = load_week2_data()
print("shape:", df.shape)
df.head()
```

## 4. Green-light — one plot

Pick **one** column that looks categorical or numeric in `df.columns` below.  
Use Copilot if you like: write a `#` comment describing what you want, then **verify** the chart makes sense.

```python
# Purpose of this cell:
# List every column name in the table (like Excel header row).
# Read this list so you can choose one column for the next plot.

print("Columns in this dataset:")
print(list(df.columns))
```

```python
# Purpose of this cell:
# Make ONE simple chart from one column — a green-light check that plotting works.
# - If the column has few categories → bar chart (counts)
# - If the column is numeric with many values → histogram
# You may change "col" to another name from the list above.
# Association in a chart is not proof of causation.

import matplotlib.pyplot as plt

col = "support_info"  # change if needed

# Sort the bars from 1 through 5 and give the requested x-axis words
support_counts = df[col].value_counts().sort_index()
labels = {
    1: "1: strongly oppose",
    2: "2",
    3: "3",
    4: "4",
    5: "5: strongly agree",
}
order = [1, 2, 3, 4, 5]
plot_labels = [labels.get(i, str(i)) for i in order]
counts = [support_counts.get(i, 0) for i in order]

fig, ax = plt.subplots(figsize=(7, 4))
ax.bar(plot_labels, counts, color="steelblue")
ax.set_title(f"Counts of {col}")
ax.set_ylabel("Count")
plt.xticks(rotation=20, ha="right")
plt.tight_layout()
plt.show()
```
